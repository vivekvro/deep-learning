```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras 
import tensorflow as tf
from sklearn.metrics import classification_report,confusion_matrix
```

```python
(x_train,y_train),(x_test,y_test) = keras.datasets.cifar10.load_data()
```

```python
labels = ['airplane','automobile','bird','cat','deer','dog','frog','horse','ship','truck']
labels
```

```python
def imshow():
    dataset=str(input("train or test"))
    index=int(input("Enter Index"))
    if dataset.lower() == 'train':
        plt.figure(figsize=(2,2))
        plt.title(f"Label : {labels[y_train[index]]}")
        plt.imshow(x_train[index])
        plt.show()
    elif dataset.lower() == 'test':
        plt.figure(figsize=(2,2))
        plt.title(f"Label : {labels[y_test[index]]}")
        plt.imshow(x_test[index])
        plt.show()
    else:
        print("wrong input")
```

```python
y_train = y_train.reshape(-1,)
```

```python
y_test = y_test.reshape(-1,)
```

```python
y_test
```

```python
x_train = x_train/255
x_test = x_test/255
```

```python
# imshow()
```

```python
ann = keras.Sequential([
    keras.layers.Flatten(input_shape=(32,32,3)),
    keras.layers.Dense(3000,activation='relu'),
    keras.layers.Dense(1000,activation='relu'),
    keras.layers.Dense(10,activation='sigmoid')
])

ann.compile(optimizer='sgd',loss='sparse_categorical_crossentropy',metrics=['accuracy'])

ann.fit(x_train,y_train,epochs=5)
```

```python
ann.evaluate(x_test,y_test)
```

```python
cnn = keras.Sequential([
    keras.layers.Conv2D(filters=32,activation='relu',kernel_size=(3,3),input_shape=(32,32,3)),
    keras.layers.MaxPooling2D((2,2)),
    keras.layers.Conv2D(filters=64,activation='relu',kernel_size=(3,3)),
    keras.layers.MaxPooling2D((2,2)),
    keras.layers.Flatten(),
    keras.layers.Dense(64,activation='relu'),
    keras.layers.Dense(10,activation='sigmoid')
])

cnn.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
cnn.fit(x_train,y_train,epochs=10)
```

```python
cnn.evaluate(x_test,y_test)
```
